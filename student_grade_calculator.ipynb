# Student Grade Calculator

## Objective
Create a Python program to assign grades
to students based on their marks.

## Concepts
- if, elif, and else
- Comparison operators
- Functions
- Lists and dictionaries
- Basic input validation

```python
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"
```

```python
students = [
    {"name": "Rahul", "marks": 95},
    {"name": "Priya", "marks": 85},
    {"name": "Aman", "marks": 75},
    {"name": "Sneha", "marks": 65},
    {"name": "Rohit", "marks": 42}
]
```

```python
for student in students:
    marks = student["marks"]

    if 0 <= marks <= 100:
        grade = calculate_grade(marks)
        print(
            f"Name: {student['name']}, "
            f"Marks: {marks}, Grade: {grade}"
        )
    else:
        print(
            f"Invalid marks for {student['name']}: {marks}"
        )
```

```python
test_marks = [100, 90, 89, 80, 79, 70, 69, 60, 59, 0]

for mark in test_marks:
    print(f"Marks: {mark} -> Grade: {calculate_grade(mark)}")
```
